## Section B - Q4: Write a query to find the day with the highest number of transactions.

```python
# === Section B - Q4: Find the day with the highest number of transactions (Databricks SQL) ===
import warnings
warnings.filterwarnings("ignore")  # Silence Python FutureWarnings so output stays clean
from pyspark.sql import SparkSession

# 1. Start our Spark session to run Databricks / Spark SQL queries
spark = (
    SparkSession.builder
    .appName("BrewBeats_SectionB_Q4")
    .master("local[*]")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("ERROR")

# 2. Load 'coffee_sales.csv' and register it as a temporary SQL table named 'coffee_sales'
coffee_df = spark.read.csv("coffee_sales.csv", header=True, inferSchema=True)
coffee_df.createOrReplaceTempView("coffee_sales")

# 3. Write our Databricks SQL query for Section B - Q4:
#    - SELECT date: picks the calendar date of the transactions
#    - DATE_FORMAT(date, 'EEEE') AS day_of_week: gets the weekday name (e.g., 'Friday', 'Monday')
#    - COUNT(*) AS total_transactions: counts how many coffee sales occurred on each date
#    - ROUND(SUM(money), 2) AS daily_revenue: calculates the total revenue earned on that date
#    - GROUP BY date: groups rows by date so we get 1 summary row per calendar day
#    - ORDER BY total_transactions DESC, daily_revenue DESC: orders from busiest day to least busy
#    - LIMIT 1: keeps only the #1 day with the highest number of transactions
q4_query = """
SELECT
    date,
    DATE_FORMAT(date, 'EEEE') AS day_of_week,
    COUNT(*) AS total_transactions,
    ROUND(SUM(money), 2) AS daily_revenue
FROM coffee_sales
GROUP BY date
ORDER BY total_transactions DESC, daily_revenue DESC
LIMIT 1
"""

print("=== Section B - Q4: Day with the Highest Number of Transactions ===")
spark.sql(q4_query).show(truncate=False)

# Bonus: Also show the Top 5 busiest days for extra context
q4_top5_query = q4_query.replace("LIMIT 1", "LIMIT 5")
print("Top 5 Busiest Days (Context):")
spark.sql(q4_top5_query).show(truncate=False)
```
