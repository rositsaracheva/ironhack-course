## Section B - Q2: Write a query to find total revenue by payment type (`cash_type`).

```python
# === Section B - Q2: Total revenue by payment type (cash_type) in Databricks SQL ===
import warnings
warnings.filterwarnings("ignore")  # Silence Python FutureWarnings so output stays clean
from pyspark.sql import SparkSession

# 1. Start our Spark session to run Databricks / Spark SQL queries
spark = (
    SparkSession.builder
    .appName("BrewBeats_SectionB_Q2")
    .master("local[*]")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("ERROR")

# 2. Load 'coffee_sales.csv' and register it as a temporary SQL table named 'coffee_sales'
coffee_df = spark.read.csv("coffee_sales.csv", header=True, inferSchema=True)
coffee_df.createOrReplaceTempView("coffee_sales")

# 3. Write our Databricks SQL query for Section B - Q2:
#    - SELECT cash_type: picks the payment method column ('card' or 'cash')
#    - COUNT(*) AS total_transactions: counts how many transactions used each payment method
#    - ROUND(SUM(money), 2) AS total_revenue: adds up all revenue ('money') per payment type, rounded to 2 decimals
#    - ROUND(AVG(money), 2) AS avg_transaction_value: calculates the average ticket size per payment type
#    - GROUP BY cash_type: groups rows by 'card' and 'cash' so totals are calculated for each payment type
#    - ORDER BY total_revenue DESC: orders from highest revenue to lowest revenue
q2_query = """
SELECT
    cash_type,
    COUNT(*) AS total_transactions,
    ROUND(SUM(money), 2) AS total_revenue,
    ROUND(AVG(money), 2) AS avg_transaction_value
FROM coffee_sales
GROUP BY cash_type
ORDER BY total_revenue DESC
"""

print("=== Section B - Q2: Total Revenue by Payment Type (cash_type) ===")
spark.sql(q2_query).show(truncate=False)
```
