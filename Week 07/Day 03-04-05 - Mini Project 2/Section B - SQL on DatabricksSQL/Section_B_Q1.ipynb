## Section B - Q1: Write a query to find the top 5 best-selling coffee types by revenue.

```python
# === Section B - Q1: Top 5 best-selling coffee types by revenue (Databricks SQL) ===
import warnings
warnings.filterwarnings("ignore")  # Silence Python FutureWarnings so output stays clean
from pyspark.sql import SparkSession

# 1. Start our Spark session (Databricks uses Spark SQL under the hood to run SQL queries)
spark = (
    SparkSession.builder
    .appName("BrewBeats_SectionB_DatabricksSQL")
    .master("local[*]")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("ERROR")

# 2. Load 'coffee_sales.csv' and register it as a SQL table called 'coffee_sales'
#    createOrReplaceTempView("coffee_sales") lets us write normal SQL queries (SELECT ... FROM coffee_sales)
#    using spark.sql(), exactly like querying a table in Databricks SQL!
coffee_df = spark.read.csv("coffee_sales.csv", header=True, inferSchema=True)
coffee_df.createOrReplaceTempView("coffee_sales")

# 3. Write our Databricks SQL query for Section B - Q1:
#    - SELECT coffee_name: picks the coffee drink name
#    - COUNT(*) AS total_transactions: counts how many cups of that coffee were sold
#    - ROUND(SUM(money), 2) AS total_revenue: adds up all sales ('money') for each coffee, rounded to 2 decimals
#    - GROUP BY coffee_name: groups the rows by coffee type so SUM() and COUNT() are calculated per drink
#    - ORDER BY total_revenue DESC: orders from highest revenue to lowest revenue
#    - LIMIT 5: keeps only the top 5 best-selling coffees
q1_query = """
SELECT
    coffee_name,
    COUNT(*) AS total_transactions,
    ROUND(SUM(money), 2) AS total_revenue
FROM coffee_sales
GROUP BY coffee_name
ORDER BY total_revenue DESC
LIMIT 5
"""

print("=== Section B - Q1: Top 5 Best-Selling Coffee Types by Revenue ===")
spark.sql(q1_query).show(truncate=False)
```
