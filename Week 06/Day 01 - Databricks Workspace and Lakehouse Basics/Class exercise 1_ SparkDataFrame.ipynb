```python
df = spark.read.option("header", True).csv("/Volumes/workspace/default/sparkdataframeexercise1/customers.csv")

df.printSchema()
df.show()
```

```python
df.select("COUNTRY", "CITY", "CUSTOMERNAME", "TERRITORY").show(5)
```

```python
spark.read.option("header", True) \
    .csv("/Volumes/workspace/default/sparkdataframeexercise1/customers.csv") \
    .createOrReplaceTempView("customer")

spark.sql("SELECT COUNTRY, CITY, CUSTOMERNAME, TERRITORY FROM CUSTOMER").show(5)
```

```python
# Filter Rows (DataFrame API)
df.select("COUNTRY", "CITY", "CUSTOMERNAME", "TERRITORY") \
    .where("COUNTRY == 'USA'") \
    .show(5)

# Filter Rows using SQL
spark.sql("""
    SELECT COUNTRY, CITY, CUSTOMERNAME, TERRITORY FROM customer
    WHERE COUNTRY = 'USA'
""").show(5)
```

```python
# Order By (DataFrame API)
df.select("COUNTRY", "CITY", "CUSTOMERNAME", "TERRITORY") \
    .where("COUNTRY in ('USA', 'France', 'Spain')") \
    .orderBy("COUNTRY") \
    .show(10)

# Order By in SQL
spark.sql("""
    SELECT COUNTRY, CITY, CUSTOMERNAME, TERRITORY FROM customer
    WHERE COUNTRY in ('USA', 'France', 'Spain')
    ORDER BY COUNTRY
""").show(10)
```

```python
df.groupBy("COUNTRY").count().show()
```
